```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/salary.csv')
display(df.head())
```

```python
print(df.columns[6])

df.rename(columns={df.columns[6]:'Salary'},inplace=True)
display(df.head())
```

```python
df.dtypes
```

```python
df['Salary']=df['Salary'].replace(',','',regex=True)
df['Salary']=df['Salary'].astype(int)
df.head
```

```python
df.info()
```

```python
df.rename(columns={df.columns[7]:'Compensation'},inplace=True)
display(df.head())

print("Sum of null:",df['Compensation'].isna().sum())
```

```python
df.rename(columns={df.columns[5]:'Context'},inplace=True)
df.rename(columns={df.columns[9]:'Other currency'},inplace=True)
df.rename(columns={df.columns[10]:'Salary context'},inplace=True)

df.drop(
    columns=['Context','Other currency','Salary context'],
)

df.head()
```

```python
#If you want to delete rows with missing values
df_clean = df.dropna()

#if you want to see the dropped rows
# ~ this symbol will find the indexed not in df_clean
dropped_rows = df[~df.index.isin(df_clean.index)]

print("Deleted/Dropped Rows:")
print(dropped_rows)
```

```python
df.dropna(thresh=2)

#Remove rows only if all their values are missing
df.dropna(how='all')

#Compare number of rows before and after
rows_before=len(df)

df_clean = df.dropna()

rows_after = len(df_clean)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

```python
#import the library
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='most_frequent')

#make a copy of your df
df_imputed= df.copy()

for column in df_imputed.columns:
  if df_imputed[column].isna().any():
     df_imputed[[column]]=imputer.fit_transform(
         df_imputed[[column]]
     )
#so your imputed column is now filled with the most common value
print(df_imputed.isna().sum())
```
